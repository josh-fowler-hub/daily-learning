## 4. Checks and interpretation

Your sentence appears after execution; comments do not appear as extra output; restart-and-run produces the same result. No equations or numerical methods are needed.

Write why each check is appropriate. Use only taught constructs; plain calculations and written checks are sufficient.

````julia
# You will see that checks implemented in the notebook, here is an example of a simple check:

x = [1, 2, 3, 4, 5, -1]  # This is a vector of numbers.

function ensure_positivity_sqrt(x)
    if x < 0
        error("x must be non-negative")
    else
        return sqrt(x)
    end
end

for i in x
    println("The square root of $i is $(ensure_positivity_sqrt(i))")
end
````

## 2. Concept: percent means per hundred

A ratio of 0.12 is 12 per hundred, written 12%. To convert a ratio to a percent number, multiply by 100; to convert a percent number back, divide by 100.

Let old be a positive reference value and new be the comparison value, expressed in the same units. Signed change = new - old. Fractional change = (new - old)/old. Signed percent change = 100*(new - old)/old. Positive means increase; negative means decrease. The change retains the quantity units, but the ratio has no physical units.

The reference is a chosen baseline, not necessarily a true or exact value. This lesson uses positive lengths, speeds and durations. With a zero baseline this formula is undefined. Do not use percentages of Celsius temperatures or arbitrary coordinate positions: their zero points do not support this interpretation.

In Python, use parentheses around the subtraction. The % operator does NOT mean percent; use *100 and /100. Keep the percent symbol in written labels, not arithmetic expressions.

## 3. Separate worked example

A ribbon changes from 40 cm to 50 cm. The signed change is 10 cm; its fraction of the old length is 10/40=0.25; the increase is 25%. The new ribbon is 125% of its old length, not 125% longer.

Reversing the comparison changes the baseline: from 50 cm to 40 cm gives 100*(40-50)/50=-20%, a 20% decrease. The changes have equal magnitudes in cm, but their denominators differ.

To apply a signed percentage p to a baseline, use new = old*(1+p/100). For the ribbon: 40*(1+25/100)=50 cm. This is a supplied arithmetic rule; no equation solving is needed. Example code to reproduce separately: ribbon_change_percent = 100*(50-40)/40.

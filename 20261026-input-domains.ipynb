## 2. Concept and worked example

A domain is a set of allowed inputs. An expression may produce a number even when the input has no meaning for its model. Physical assumptions can restrict inputs more than arithmetic does.

Separate example: a fictional cart has position p(t)=3+2*t metres for elapsed time from 0 through 4 seconds, including both endpoints. The expression evaluates at t=-1 and t=5, but those times are outside the supplied model interval. At t=2 it gives 7 m. A finite table only samples the interval; it does not list every allowed input.

For an average speed d/t, zero elapsed time is excluded by division; negative elapsed time is excluded by this elapsed-time interpretation. State constraints in words today. Comparisons and conditionals will be taught next.

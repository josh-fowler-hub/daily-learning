## 6. Optional stretch

Choose a different consistent unit system and recover the same physical distance.

````julia
printlnmd("### Optional Exercise Answer")

printlnmd("We'll start with a distance of 25 miles.")

miles_to_ft(distance) = distance*5280
ft_to_m(distance) = distance*0.3048
m_to_km(distance) = distance/1000
km_to_mi(distance) = distance*0.621371

distance_mi1 = 25 # miles
distance_ft = miles_to_ft(distance_mi1)
distance_m = ft_to_m(distance_ft)
distance_km = m_to_km(distance_m)
distance_mi2 = km_to_mi(distance_km)

printlnmd("Let's see if the following code produces the same result after converting several times:

```julia
miles_to_ft(distance) = distance*5280
ft_to_m(distance) = distance*0.3048
m_to_km(distance) = distance/1000
km_to_mi(distance) = distance*0.621371

distance_mi1 = 25 # miles
distance_ft = miles_to_ft(distance_mi1)
distance_m = ft_to_m(distance_ft)
distance_km = m_to_km(distance_m)
distance_mi2 = km_to_mi(distance_km)
```
")

printlnmd("distance_mi1 = $distance_mi1 [mi] → distance_ft = $distance_ft [ft] → distance_m = $distance_m [m] → distance_km = $distance_km [km] → distance_mi2 = $distance_mi2 [mi]")
printlnmd("This is an important result that we will discuss later, for now just consider them equal and the exercise passed.")
````

## 1. Concept and notation

A quantity describes something measurable, a symbol names it, a value gives its size, and a unit supplies the scale. For example, t may denote time, 4 is a numerical value, and s means seconds. ## 4. Checks and interpretation

Every example identifies quantity, symbol, value and unit; invented data are labeled fictional; no assumption that a symbol uniquely determines its physical meaning.

Write why each check is appropriate. Use only taught constructs; plain calculations and written checks are sufficient.

````julia
import Base: @doc

display(disty, "### Example")

"""
    valid_cooling(Tw, Tb)

Check whether a valid cooling configuration exists.

# Arguments
- Tw :: Float64 => Wall Temperature
- Tb :: Float64 => Bulk Temperature

# Outputs
- result::Bool

# Examples
```julia-repl
julia> valid_cooling(250, 275)
false
```
"""
function valid_cooling(Tw::Float64, Tb::Float64)
    if Tw <= Tb
        return false
    else
        return true
    end
end

Tw  = 250.0 # K
display(disty, "Wall Temperature: \$T_w = $Tw \\left[K\\right]\$")
Tb = 275.0 # K
display(disty, "Bulk Temperature: \$T_b = $Tb \\left[K\\right]\$")

display(disty, "We require that \$T_w \\ge T_b\$ for cooling.")
display(disty, "Let's check it with the function valid_cooling:")
display(disty, @doc valid_cooling)

result = valid_cooling(Tw, Tb)

if !result
    error("Tw must be higher than Tb for a valid cooling configuration.")
end
````

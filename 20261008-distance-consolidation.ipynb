## 1. Concept and notation

Combine taught arithmetic, units and variables. A supplied model d=v*t assumes constant speed in each segment; it does not describe acceleration or a real aircraft performance envelope.

```julia
function printlnmd(line)
    display("text/markdown", line)
end

struct Segment
    v
    t
    d

    function Segment(v, t)
        d = v*t
        new(v, t, d)
    end
end

struct Trip
    segments
    tt
    td
    vbar
    function Trip(segments)
        total_time = 0.0
        for segment in segments
            total_time += segment.t
        end

        total_distance = 0.0
        for segment in segments
            total_distance += segment.d
        end

        average_time = total_distance/total_time

        new(segments, total_time, total_distance, average_time)
    end
end
```

## 2. Worked example (different from the exercise)

Two segments at 5 m/s for 2 s and 10 m/s for 3 s cover 10+30=40 m.



```julia
segment1 = Segment(5, 2)
segment2 = Segment(10, 3)
trip = Trip([segment1, segment2])

printlnmd("The total distance covered is \$ $(trip.td)\\left[m\\right]\$")
```

## 3. Required exercise

1. Model a fictional vehicle traveling 10 m/s for 4 s, then 15 m/s for 6 s.
2. Compute each distance, total distance and total time.
3. Compute trip-average speed as total distance divided by total time.

```julia
segment1 = Segment(10.0, 4.0)
printlnmd("The total distance traveled in segment 1 is \$ $(segment1.d) \\left[m\\right]\$")

segment2 = Segment(15.0, 6.0)
printlnmd("The total distance traveled in segment 2 is \$ $(segment2.d) \\left[m\\right]\$")

trip = Trip([segment1, segment2])
printlnmd("The total distance traveled during the trip is \$ $(trip.td) \\left[m\\right]\$")
printlnmd("The total time traveled during the trip is \$ $(trip.tt) \\left[s\\right]\$")
printlnmd("The average speed traveled during the trip is \$ $(trip.vbar) \\left[m/s\\right]\$")
```

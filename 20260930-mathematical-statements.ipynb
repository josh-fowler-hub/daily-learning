## 3. Required exercise

1. Classify "7 equals 7", "3 is greater than 8", and "Find the height" as true statement, false statement, or instruction.
2. Write the symbols =, <, and > in Markdown and explain each.
3. Give a simple equality and an ordering statement of your own.

````julia
struct statement
    text::String
    truth_value::Union{Bool, Nothing}
end

evaluations = (
    "7 equals 7" => statement("7 equals 7", true),
    "3 is greater than 8" => statement("3 is greater than 8", false),
    "Find the height of the building" => statement("Find the height of the building", nothing)
)

for (evaluation, result) in evaluations
    result = result.truth_value === nothing ? "not a statement" : (result.truth_value ? "true" : "false")
    if result == "not a statement"
        println("$evaluation is $result.")
    else
        println("$evaluation is a statement and it is $result.")
    end
end
````

This is how you write LaTeX in Markdown:

1. $=$: \$=\$
2. $\lt$: \$\lt\$
3. $\gt$: \$\gt\$
4. $\le$: \$\le\$
5. $\ge$: \$\ge\$

## 4. Checks and interpretation

````julia
# Here is another example of a check.

function checkindex(index::Int, array::Vector{Int})
    if index < 1 || index > length(array)
        error("Index $index is out of bounds for array of length $(length(array)).")
        return nothing
    else
        return "Index is within bounds."
    end
end

bvec = [1, 2, 3, 4, 5]

for i in 1:6
    str = checkindex(i, bvec)
    if str !== nothing
        println(str)
        continue
    end
end

````

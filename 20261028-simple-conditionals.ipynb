## 2. Concept and separate worked example

A comparison produces a Boolean. A conditional uses that Boolean to choose which statements execute. Python checks an if condition first. If false, it checks the next elif condition. The first true branch runs; otherwise else runs. Colons introduce branches. Indent each branch body consistently, using four spaces. Only the selected branch body executes.

Separate example: a fictional container accepts integer counts from 0 through 4. The following example classifies a count; it is not the assigned exercise.

count = 2
if count < 0:
    print("Negative count")
elif count > 4:
    print("Too many")
else:
    print("Within stated range")

With count=2, both tests are false, so else runs. With count=-1, the first branch runs. With count=5, the second runs. At 0 and 4, else runs because the bounds are inclusive. Two separate if statements are not the same as one mutually exclusive if/elif/else chain.

A calculation placed after the whole chain runs regardless of which branch ran. To restrict a model calculation, place it inside the accepted branch. Assign outputs afresh on each run or keep them local to a function so an old result cannot be mistaken for a new one.

## 1. Concept and notation

Arithmetic expressions represent a precise sequence of operations. Python uses ** for powers and parentheses to group terms. Julia uses ^ for powers. A number alone has no physical unit.

````julia
using Plots
gr()

disty = "text/markdown"

display(disty, "## Arithmetic")
display(disty, "Here is an example of how to use arithmetic.")

f(x) = 0.659*x^2 + 34.4*x - 65.4
xs = LinRange(0, 10, 50)
ys = [f(x) for x in xs]

display(disty, "
```julia
f(x) = 0.659*x^2 + 34.4*x - 65.4
xs = LinRange(0, 10, 50)
ys = [f(x) for x in xs]
```
")

plot(xs, ys, lw=2, lc=:black, label="Function Value")
title!("\$ f\\left(x\\right) = 0.659x^2 + 34.4x - 65.4\$")
xlabel!("X-Value")
ylabel!("Y-Value")
````

## 6. Optional stretch

Write two different expressions with the same numerical result and explain their operations.

````julia
f1(x) = 3x + 4
f2(x) = -x + 8

println("When x=1, we have f1 = $(f1(1)), and f2 = $(f2(1))")
println("This works because the two functions are not parallel and intersect at x = 1\n You can see from the plot below:\n")

xs = LinRange(-10, 10, 100)
y1s = [f1(x) for x in xs]
y2s = [f2(x) for x in xs]

plot(xs, y1s, lw=2, lc=:black, label="\$f_1\\left(x\\right)\$")
plot!(xs, y2s, lw=2, lc=:red, label="\$f_2\\left(x\\right)\$")
plot!(grid=true)
title!("Comparison of \$f_1\\left(x\\right)\$ and \$f_2\\left(x\\right)\$")
xlabel!("X")
ylabel!("Y")
````

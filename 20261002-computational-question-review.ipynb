## 4. Checks and interpretation

Inputs and output are distinct; both inputs carry consistent units; the assumption is explicit; the notebook reproduces its output. This reviews orientation, not evidence of mastery of later mathematics.

Write why each check is appropriate. Use only taught constructs; plain calculations and written checks are sufficient.

````julia
function pressure_positivity(p)
    index = 1
    for pressure in p
        if pressure <= 0.0
            error("The pressure is non-positive at index $index. Exiting.")
        end
        index += 1
    end
end

pressures = [1, 2, 3, 5, 8, 10, 7, 6, 3, 2, 1, 0]

pressure_positivity(pressures)
````

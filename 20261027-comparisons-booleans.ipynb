## 2. Concept and separate worked example

A comparison asks a question and returns True or False, called a Boolean value. Python uses < for less than, <= for less than or equal, > for greater than, >= for greater than or equal, == for equal, and != for not equal. A single = assigns a value; == compares values. Use matching physical units.

Separate example: a fictional shelf allows a mass no greater than 12 kg. With mass_kg=9, mass_kg <= 12 returns True. At mass_kg=12 it still returns True; mass_kg < 12 returns False. Inclusion of the endpoint changes the answer. Store an answer with within_limit = mass_kg <= 12, then print(within_limit). This stores a Boolean, not a new mass.

## 2. Powers of ten, taught from the beginning

An exponent counts repeated multiplication: 10^3 means 10*10*10 = 1000. In this lesson the base is always 10. We define 10^0 = 1. A negative exponent means a reciprocal: 10^(-1)=1/10=0.1 and 10^(-3)=1/1000=0.001. A negative exponent does not make the number negative.

Multiplying by 10 moves the decimal point one place to the right; dividing by 10 moves it one place to the left. Scientific notation writes a positive number as a times 10^n, where the coefficient a is at least 1 and less than 10, and n is an integer (whole number, possibly negative or zero). This is called normalized scientific notation. The powers supply the scale; the coefficient supplies the remaining value. Zero is a separate case and is outside today's normalization tasks.

In Python use ** for powers: 10**3 and 10**(-3). The ^ operator is not exponentiation in Python. A compact number literal such as 2.4e3 means 2.4*10**3. Here e is part of numeric notation, not a variable or the exponential function. 2.4e-3 means 2.4*10**(-3). The unit still belongs in the variable name or written label.

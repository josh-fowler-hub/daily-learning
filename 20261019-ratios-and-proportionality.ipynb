## 2. New idea and separate worked example

A ratio compares by division. The ratio of A to B is A/B, so order matters and B cannot be zero. For comparable quantities, convert to the same units first: 1 m / 50 cm = 100 cm / 50 cm = 2. The units cancel. A rate compares different kinds of quantities; distance/time retains units of speed.

Direct proportionality means y = k*x with one constant k. For nonzero x, y/x = k. Multiplying x by a factor multiplies y by the same factor. Its graph is a straight line through the origin. A straight line alone is insufficient: y = b + k*x with nonzero b is not directly proportional.

Worked example, separate from your assignment: identical cards each have mass 3 g. For counts 1, 2, and 4, total mass is 3, 6, and 12 g. The three mass/count ratios are all 3 g per card. Doubling the count doubles mass. Adding a fixed 2 g envelope gives packaged mass 5, 8, and 14 g: mass/count is no longer constant. The envelope changes the intercept, not the mass added per card.

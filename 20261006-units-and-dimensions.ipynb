## 1. Concept and notation

A physical quantity is a numerical value paired with a unit. Add only compatible dimensions. Multiplying speed (m/s) by time (s) gives distance (m). Conversion factors are ratios equal to one.

```julia
speed = 200 # m/s
t = 300 # s

distance = speed*t
m_to_km(d) = d/1000 # 1000 m is equal to 1 km

println("The distance traveled is $distance m, or $(m_to_km(distance)) km.")
```

## 2. Worked example (different from the exercise)

A speed of 20 m/s sustained for 3 s gives 60 m. A length of 2 km is 2*1000 = 2000 m.
